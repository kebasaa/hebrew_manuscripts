# PDF to OSIS

Thin notebook wrapper around the tested `pdf2osis` package. Run this notebook with the `tp` Conda environment.

```python
from pathlib import Path
from pdf2osis import convert_pdf, get_profile

ROOT = Path.cwd().parent if Path.cwd().name == "python" else Path.cwd()
SOURCE = ROOT / "data" / "00_source_files"
OUTPUT = ROOT / "data" / "01_osis"
```

```python
reports = []
for book in ("rev", "jas"):
    profile = get_profile(book)
    reports.append(convert_pdf(profile.default_path(SOURCE), profile, OUTPUT))
reports
```
